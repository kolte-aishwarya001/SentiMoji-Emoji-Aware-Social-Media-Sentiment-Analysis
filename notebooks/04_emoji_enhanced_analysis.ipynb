```python
import pandas as pd
import emoji

df = pd.read_csv(
    "../dataset/emoji_enhanced_train.csv",
    encoding="utf-8"
)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nFirst 10 records:")
print(df.head(10))
```

```python
def count_emojis(text):
    return len(emoji.emoji_list(str(text)))


df["emoji_count"] = df["text"].apply(count_emojis)

total_records = len(df)
emoji_records = (df["emoji_count"] > 0).sum()
total_emojis = df["emoji_count"].sum()

print("Total records:", total_records)
print("Records containing emojis:", emoji_records)
print("Total emojis:", total_emojis)

print(
    "Emoji coverage:",
    round((emoji_records / total_records) * 100, 2),
    "%"
)
```

```python
emoji_by_sentiment = pd.crosstab(
    df["sentiment"],
    df["emoji_count"] > 0
)

print(emoji_by_sentiment)
```

```python
emoji_records_df = df[df["emoji_count"] > 0]

print("\nEmoji tweets by sentiment:")
print(
    emoji_records_df["sentiment"].value_counts()
)
```
